```python
import pyspark.sql.functions as F
```

```python
file_name1 ="sampledatasuperstore.parquet"
file_name2 = "sampledata - superstorexref.parquet"
container2 = "silver"

dataframe1 = spark.read.parquet(f"abfss://{container2}@{account_name}.dfs.core.windows.net/almost_resolved/{file_name1}")
dataframe2 = spark.read.parquet(f"abfss://{container2}@{account_name}.dfs.core.windows.net/almost_resolved/{file_name2}").dropDuplicates(["Order ID"])

dataframe1.display()
display(dataframe2)
```

```python
dataframe1_final = dataframe1.withColumnRenamed("Order ID", "OrderID")
dataframe2_final = dataframe2.withColumnRenamed("Order ID", "OrderID")
dataframe1_final.display()
```

```python
sample_data_superstore = dataframe1_final.join(
    dataframe2_final,
    on= "OrderID",
    how= "left"
).withColumnRenamed("Customer Return", "CustomerReturn"
).withColumnRenamed("Ship Date", "ShipDate"
).withColumnRenamed("Order Date", "OrderDate")
sample_data_superstore.display()
```

```python
sample_data_superstore_trans = sample_data_superstore.filter(
    F.col("CustomerReturn")== "No"
    ).withColumn(
    "Profit1", F.round(F.col("Profit"),2)
).withColumn(
    "Todays_Date", F.current_date()
).withColumn(
    "Order_Quantity_Flag", F.when(
        F.col("Quantity") >= 5, "Huge Order"
    ).otherwise("SmallOrder")
)
sample_data_superstore_trans.display()
```

```python
sample_data_superstore.createOrReplaceTempView("sample_data_superstore_vw")

sample_data_superstore_sql = spark.sql(
    """
    select *,
    ROUND(Profit) AS RoundedProfit,
    current_date() AS TodayDate,
    CASE
        WHEN Quantity >= 5 THEN 'Huge Order'
        ELSE 'Small Order'
    END AS Order_Quantity_Flag
    from sample_data_superstore_vw
    where CustomerReturn = "No"

    """
)
sample_data_superstore_sql.display()
```

```python
sample_data_superstore_final = sample_data_superstore_sql.withColumn(
    "DateDiff", F.datediff(F.col("ShipDate"), F.col("OrderDate"))
)

sample_data_superstore_final.display()
```
